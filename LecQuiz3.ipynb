````python
import re
import pandas as pd
import numpy as np

df_clean = df.copy()

def clean_name(x):
    if pd.isna(x):
        return np.nan
    x = x.strip()
    x = re.sub(r'(?<=[a-z])(?=[A-Z])', ' ', x)  # delaCruz -> dela Cruz
    return x.title()

df_clean['patient_name'] = df_clean['patient'].apply(clean_name)
df_clean['patient_name']
````

````python
def clean_sex(x):
    x = str(x).strip().lower()
    if x in ('m', 'male'):
        return 'M'
    elif x in ('f', 'female'):
        return 'F'
    return np.nan

df_clean['sex'] = df_clean['gender'].apply(clean_sex)
df_clean['sex'].value_counts()
````

````python
def clean_date(x):
    if pd.isna(x):
        return pd.NaT
    x = str(x).strip()
    for fmt in ('%Y-%m-%d', '%m/%d/%Y', '%d-%m-%Y', '%Y/%m/%d'):
        try:
            return pd.to_datetime(x, format=fmt)
        except (ValueError, TypeError):
            continue
    return pd.to_datetime(x, errors='coerce')

df_clean['visit_date'] = df_clean['date_visited'].apply(clean_date)
df_clean.loc[4, 'visit_date'] = pd.Timestamp('2026-01-08')
df_clean.loc[9, 'visit_date'] = pd.Timestamp('2026-01-12')

df_clean['visit_date']
````

````python
def clean_age(x):
    if pd.isna(x):
        return np.nan
    return np.nan if (x < 0 or x > 110) else x
df_clean['age'] = df_clean['patient_age'].apply(clean_age)

def clean_weight(x):
    if pd.isna(x):
        return np.nan
    x = str(x).lower().replace('kg', '').strip()
    try:
        return float(x)
    except ValueError:
        return np.nan
df_clean['weight_kg'] = df_clean['weight'].apply(clean_weight)
df_clean.loc[df_clean['weight_kg'] < 0, 'weight_kg'] = np.nan

df_clean['height_cm'] = df_clean['height']
df_clean.loc[df_clean['height_cm'] < 50, 'height_cm'] = np.nan

df_clean['bp_systolic'] = pd.to_numeric(df_clean['systolic_bp'], errors='coerce')
df_clean.loc[(df_clean['bp_systolic'] < 60) | (df_clean['bp_systolic'] > 250), 'bp_systolic'] = np.nan

df_clean['temperature_c'] = df_clean['temperature']
df_clean.loc[df_clean['temperature_c'] > 42.0, 'temperature_c'] = np.nan

df_clean[['age','weight_kg','height_cm','bp_systolic','temperature_c']].describe()
````

````python
def clean_diagnosis(x):
    # Handle NaN values explicitly
    if pd.isna(x):
        return np.nan
    x = str(x).strip().lower()
    mapping = {
        'hypertension': 'Hypertension', 'common cold': 'Common Cold',
        'diabetes': 'Diabetes', 'diabetes type 2': 'Diabetes',
        'flu': 'Flu', 'asthma': 'Asthma',
        'migraine': 'Migraine', 'migrain': 'Migraine', 'uti': 'UTI',
    }
    return mapping.get(x, x.title())

df_clean['diagnosis'] = df_clean['illness'].apply(clean_diagnosis)
df_clean['diagnosis'].value_counts()
````

````python
def clean_followup(x):
    return str(x).strip().lower() in ('yes', 'y')

df_clean['follow_up_needed'] = df_clean['needs_followup'].apply(clean_followup)
df_clean['follow_up_needed'].value_counts()
````

````python
df_clean.loc[df_clean['weight_kg'] > 140, 'weight_kg'] = np.nan

dedup_keys = ['patient_name', 'age', 'sex', 'visit_date']
df_final = df_clean.drop_duplicates(subset=dedup_keys, keep='first').copy()
df_named = df_final.dropna(subset=['patient_name'])
df_final
````
